```python
# generate data.yaml

import yaml

data = {
    'path': '/content/datasets/dataset_player_250508',
    'train': 'train',
    'val': 'valid',
    'nc': 4,
    'names': ['ball', 'court', 'net', 'player']
}

with open('data.yaml', 'w') as file:
    yaml.dump(data, file)
```

```python
from ultralytics import YOLO

model = YOLO('yolov8n.pt')
dataset_yaml = 'data.yaml'

model.train(data=dataset_yaml,
            task='detect',
            epochs=150,
            imgsz=640,
            batch=16,
            project='YOLOv11-player',
            name=f'EX1',
            device=0,
            )
```
