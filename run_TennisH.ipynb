```python
# Install CUDA 11.8
!wget https://developer.download.nvidia.com/compute/cuda/11.8.0/local_installers/cuda_11.8.0_520.61.05_linux.run
!chmod +x cuda_11.8.0_520.61.05_linux.run
!./cuda_11.8.0_520.61.05_linux.run --silent --toolkit

# Set environment variables
import os
os.environ['CUDA_HOME'] = '/usr/local/cuda-11.8'
os.environ['PATH'] = '/usr/local/cuda-11.8/bin:' + os.environ['PATH']
os.environ['LD_LIBRARY_PATH'] = '/usr/local/cuda-11.8/lib64:' + os.environ.get('LD_LIBRARY_PATH', '')
```

```python
# Verify installation
!nvcc --version
```

```python
# Download and install cuDNN 8.6.0 for CUDA 11.8
!wget https://developer.download.nvidia.com/compute/redist/cudnn/v8.7.0/local_installers/11.8/cudnn-linux-x86_64-8.7.0.84_cuda11-archive.tar.xz
!tar -xf cudnn-linux-x86_64-8.7.0.84_cuda11-archive.tar.xz

# Copy cuDNN files to CUDA directory
!cp cudnn-linux-x86_64-8.7.0.84_cuda11-archive/include/cudnn*.h /usr/local/cuda-11.8/include/
!cp cudnn-linux-x86_64-8.7.0.84_cuda11-archive/lib/libcudnn* /usr/local/cuda-11.8/lib64/

# Set permissions
!chmod a+r /usr/local/cuda-11.8/include/cudnn*.h /usr/local/cuda-11.8/lib64/libcudnn*

# Update library cache
!ldconfig
```

```python
# Check if cuDNN files are properly installed
!ls -la /usr/local/cuda-11.8/lib64/libcudnn*
!ls -la /usr/local/cuda-11.8/include/cudnn*

# Verify cuDNN version
!cat /usr/local/cuda-11.8/include/cudnn_version.h | grep CUDNN_MAJOR -A 2
```
