```python
# generate data.yaml

import yaml

data = {
    'path': '/content/datasets/dataset_score_board_250507',
    'train': 'train',
    'val': 'valid',
    'nc': 1,
    'names': ['score-board']
}

with open('data.yaml', 'w') as file:
    yaml.dump(data, file)
```

```python
from ultralytics import YOLO

model = YOLO('yolov8n.pt')
dataset_yaml = 'data.yaml'

model.train(data=dataset_yaml,
            task='detect',
            epochs=150,
            imgsz=640,
            batch=64,
            project='YOLOv11-scoreboard',
            name=f'EX1',
            device=0,
            )
```
